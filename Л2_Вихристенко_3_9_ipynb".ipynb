## 13. Лінійна алгебра

### Матричний добуток

```python
a = np.array([
    [1, 2, 3],
    [0, 1, 2]
])

b = np.array([
    [ 1, 2],
    [ 0, 4],
    [-1, 0]
])

c = a @ b
print(c)
```

### Діагональ, транспонування та визначник

```python
a = np.array([
    [ 1.,  0., 3.],
    [-1., -1., 2.],
    [ 4.,  7., 2.]
])

print("diag:", np.diag(a))
print("\nA.T =\n", a.T)

det = np.linalg.det(a)
print("\ndet(A) =", det)
print("det близький до -25:", np.isclose(det, -25.0))
```

### Розв’язання системи `Ax = b`

```python
a = np.array([
    [1.,  2.,  5.],
    [1., -1.,  3.],
    [3., -6., -1.]
])

b = np.array([-9., 2., 25.])

x = np.linalg.solve(a, b)

print("x =", x)
print("Перевірка A @ x ≈ b:", np.allclose(a @ x, b))
```

Для розв’язання системи не потрібно явно обчислювати `A⁻¹`.  
У числових обчисленнях перевага надається `np.linalg.solve(A, b)`.


### Метод найменших квадратів

```python
x = np.array([
    [1., 1.],
    [1., 2.],
    [1., 3.],
    [1., 4.]
])

y = np.array([2.1, 3.9, 6.2, 8.1])

beta, residuals, rank, s = np.linalg.lstsq(x, y, rcond=None)

print("beta:", beta)
print("residuals:", residuals)
print("rank:", rank)
print("singular values:", s)
```

### Власні значення та власні вектори

```python
a = np.array([
    [2., 1.],
    [1., 2.]
])

values, vectors = np.linalg.eigh(a)

print("Власні значення:", values)
print("Власні вектори:\n", vectors)
```

### SVD

```python
a = np.array([
    [3., 1.],
    [1., 3.],
    [1., 1.]
])

U, s, Vt = np.linalg.svd(a, full_matrices=False)

print("U =\n", U)
print("\ns =", s)
print("\nVt =\n", Vt)
```

## 18. Узагальнюючий приклад: підготовка даних і лінійний прогноз

```python
# 1. Вхідні дані: два фактори та цільова змінна
x = np.array([
    [10., 2.],
    [12., 3.],
    [15., 4.],
    [18., 5.],
    [20., 6.]
])

y = np.array([21., 25., 31., 36., 40.])

# 2. Фільтрація
mask = np.isfinite(x).all(axis=1) & np.isfinite(y)
x = x[mask]
y = y[mask]

# 3. Стандартизація ознак
mu = x.mean(axis=0)
sigma = x.std(axis=0)
x_std = (x - mu) / sigma

# 4. Додаємо стовпець одиниць для вільного члена
x_design = np.column_stack([
    np.ones(x_std.shape[0]),
    x_std
])

# 5. Оцінювання параметрів методом найменших квадратів
beta, *_ = np.linalg.lstsq(x_design, y, rcond=None)

# 6. Прогноз і MSE
y_pred = x_design @ beta
mse = np.mean((y - y_pred) ** 2)

print("beta =", beta)
print("y_pred =", y_pred)
print("MSE =", mse)
```

## 19. Підсумкові тренувальні завдання

2. Обчисліть середнє та стандартне відхилення кожної ознаки.
3. Стандартизуйте матрицю через broadcasting.
4. Виберіть рядки, у яких значення першої ознаки додатне.
5. Додайте до матриці новий стовпець одиниць.
6. Створіть вектор ваг і обчисліть прогноз `X @ w`.
7. Для заданого `y_true` обчисліть MSE.
8. Перевірте, чи немає в масивах `NaN` або `inf`.


```python
# Виконайте підсумкове завдання тут
import numpy as np

rng = np.random.default_rng(42)

X = rng.standard_normal((5, 4))

mean = X.mean(axis=0)
std = X.std(axis=0)

X_scaled = (X - mean) / std

X_positive_first_feature = X[X[:, 0] > 0]

ones_column = np.ones((X.shape[0], 1))
X_with_ones = np.hstack([X, ones_column])

w = rng.standard_normal(5)
y_pred = X_with_ones @ w

y_true = rng.standard_normal(5)
mse = np.mean((y_true - y_pred) ** 2)

has_nan = np.isnan(X_scaled).any() or np.isnan(y_pred).any()
has_inf = np.isinf(X_scaled).any() or np.isinf(y_pred).any()

print("Початкова матриця X:\n", X.round(3))
print("\nСереднє за ознаками:", mean.round(3))
print("     Стандартне відхилення:", std.round(3))
print("\nСтандартизована матриця X_scaled:\n", X_scaled.round(3))
print("\nРядки, де 1-ша ознака > 0:\n", X_positive_first_feature.round(3))
print("\nМатриця зі стовпцем одиниць:\n", X_with_ones.round(3))
print("\nПрогноз y_pred:", y_pred.round(3))
print("\nMSE:", round(mse, 4))
print(f"\nНаявність NaN: {has_nan}, Наявність inf: {has_inf}")
```
