### Завдання 25

````python
# Код студента
a = np.array([1, 2])
b = np.array([2, 1])

cos_theta = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
angle_deg = np.degrees(np.arccos(cos_theta))

print("Косинус кута (cos θ):", cos_theta)
print("Кут у градусах:", angle_deg)
````

### Завдання 28
Для матриці:


````python
# Код студента
A = np.array([[4, 7],
              [2, 6]])

det_A = np.linalg.det(A)

A_inv = np.linalg.inv(A)

identity_check = A @ A_inv

print("Визначник (det):", det_A)
print("\nОбернена матриця (A_inv):\n", A_inv)
print("\nПеревірка (A @ A_inv):\n", np.round(identity_check))
````

### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
# Код студента
A = np.array([[10, 20, 30],
              [20, 30, 40],
              [30, 40, 50]])

mu = np.mean(A, axis=0)
sigma = np.std(A, axis=0)
z = (A - mu) / sigma

print(z)
````

```text
sales = [[120, 150, 170, 160],

         [200, 180, 210, 230]]
```

Рядки — товари, стовпці — місяці.

Визначте:
1. загальний обсяг продажів кожного товару;

Створіть квадратну матрицю `5 × 5` із випадкових цілих чисел від `1` до `20`.

Виконайте:
1. знайдіть суму головної діагоналі;
2. знайдіть суму побічної діагоналі;
3. знайдіть максимальний елемент кожного рядка;
4. знайдіть мінімальний елемент кожного стовпця;
5. замініть усі елементи, менші за середнє значення всієї матриці, на `0`.

````python
# Код студента
np.random.seed(42)

matrix = np.random.randint(1, 21, size=(5, 5))
print("Початкова матриця:\n", matrix)

main_diag_sum = np.trace(matrix)

anti_diag_sum = np.trace(np.fliplr(matrix))

max_per_row = np.max(matrix, axis=1)

min_per_col = np.min(matrix, axis=0)

mean_val = np.mean(matrix)
modified_matrix = matrix.copy()
modified_matrix[modified_matrix < mean_val] = 0

print("\n1. Сума головної діагоналі:", main_diag_sum)
print("2. Сума побічної діагоналі:", anti_diag_sum)
print("3. Максимальний елемент кожного рядка:", max_per_row)
print("4. Мінімальний елемент кожного стовпця:", min_per_col)
print(f"5. Матриця після заміни (середнє було {mean_val:.2f}):\n", modified_matrix)
````

### Завдання 40
Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
2. Перевірте знайдений розв’язок матричним множенням `A @ x`.
3. Порівняйте результат із вектором `b`.

````python
# Код студента
A = np.array([[1, 2, 1],
              [2, 1, -1],
              [3, -1, 2]])

b = np.array([9, 8, 3])

x = np.linalg.solve(A, b)
print("Розв'язок системи (x, y, z):", x)

check_b = A @ x
print("Результат множення A @ x:", check_b)

print("Збігається з вектором b:", np.allclose(check_b, b))
````
