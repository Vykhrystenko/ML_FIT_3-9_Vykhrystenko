## Завдання 3. Скалярний добуток і норма вектора

Для векторів `u = [3, -1, 2]` та `v = [1, 4, -2]`:

1. Обчисліть скалярний добуток.
2. Знайдіть евклідову норму кожного вектора.
3. Обчисліть косинус кута між векторами за формулою

\[
\cos(\alpha)=\frac{u\cdot v}{\|u\|\|v\|}
\]


````python
u = np.array([3, -1, 2])
v = np.array([1, 4, -2])

# Код студента

dot_product = np.dot(u, v)

norm_u = np.linalg.norm(u)
norm_v = np.linalg.norm(v)

cos_alpha = dot_product / (norm_u * norm_v)

print(f"dot_product: {dot_product}")
print(f"norm_u: {norm_u}")
print(f"norm_v: {norm_v}")
print(f"cos_alpha: {cos_alpha}")
````

## Завдання 12. Власна функція

Напишіть функцію `vector_stats(x)`, яка приймає числовий вектор і повертає:

- мінімум;
- максимум;
- середнє;
- медіану;
- стандартне відхилення.

Перевірте функцію на векторі:

`[4, 7, 1, 9, 3, 8, 5]`.

````python
def vector_stats(x):
    min_val = np.min(x)
    max_val = np.max(x)
    mean_val = np.mean(x)
    median_val = np.median(x)
    std_val = np.std(x)

    return {
        "мінімум": min_val,
        "максимум": max_val,
        "середнє": mean_val,
        "медіана": median_val,
        "стандартне відхилення": std_val
    }


test_vector = np.array([4, 7, 1, 9, 3, 8, 5])

stats = vector_stats(test_vector)

for stat_name, value in stats.items():
    print(f"{stat_name.capitalize()}: {value:.4f}")
````

## Завдання 13. Функція з умовою

Створіть функцію

\[
f(x) =
\begin{cases}
x^2, & x < 0 \\
2x + 1, & x \geq 0
\end{cases}
\]

Застосуйте її до вектора значень від `-5` до `5`.

Спробуйте реалізувати функцію двома способами:
1. через звичайну функцію Python;
2. через `np.where()`.

````python
x = np.arange(-5, 6)

# Варіант 1
def f_python(x_val):
    if x_val < 0:
        return x_val ** 2
    else:
        return 2 * x_val + 1
y_variant1 = np.array([f_python(val) for val in x])
# Варіант 2
def f_numpy(x_arr):
    return np.where(x_arr < 0, x_arr ** 2, 2 * x_arr + 1)

y_variant2 = f_numpy(x)
print("Вхідний вектор x:", x)
print("Варіант 1 (Python):", y_variant1)
print("Варіант 2 (np.where):", y_variant2)
````

## Завдання 15. Підсумкове

Згенеруйте матрицю `5 × 5` із випадкових цілих чисел від `1` до `100`.

Виконайте такі дії:
1. Знайдіть середнє значення кожного рядка.
2. Знайдіть середнє значення кожного стовпця.
3. Замініть усі елементи, менші за загальне середнє матриці, на `0`.
4. Знайдіть рядок із найбільшою сумою.
5. Нормалізуйте початкову матрицю за формулою:

\[
x_{norm} = \frac{x - x_{min}}{x_{max} - x_{min}}
\]

6. Напишіть функцію, яка виконує таку нормалізацію для довільної числової матриці.

````python
rng = np.random.default_rng(42)
matrix = rng.integers(1, 101, size=(5, 5))

print("Початкова матриця:")
print(matrix)

# Код студента
row_means = matrix.mean(axis=1)

col_means = matrix.mean(axis=0)

matrix_filtered = matrix.copy()
matrix_filtered[matrix_filtered < matrix.mean()] = 0

max_sum_row_idx = matrix.sum(axis=1).argmax()
max_sum_row = matrix[max_sum_row_idx]

def normalize_matrix(mat):
    return (mat - mat.min()) / (mat.max() - mat.min())

matrix_normalized = normalize_matrix(matrix)

print("\nСереднє по рядках:", row_means)
print("\nСереднє по стовпцях:", col_means)
print("\nМатриця після заміни елементів < mean на 0:\n", matrix_filtered)
print(f"\nРядок з найбільшою сумою (індекс {max_sum_row_idx}):", max_sum_row)
print("\nНормалізована матриця (від 0 до 1):\n", matrix_normalized.round(4))
````
